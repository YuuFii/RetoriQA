```python
def load_pdf(pdf_path):
    doc = pymupdf.open(pdf_path)
    pages = []

    for page_number, page in enumerate(doc):
        text = page.get_text().encode("utf8")
        if text.strip():
            pages.append({
                "page": page_number + 1,
                "text": text,
            })

    return pages
```

```python
pages = load_pdf("CV_Nico Arya Divano.pdf")
pages[0]
```

```python
import re

def normalisasi_teks(value):
    value = value.decode('utf-8', errors='ignore') if isinstance(value, bytes) else value
    value = re.sub(r'\s+', ' ', value).strip()
    return value
```

```python
texts = "".join(normalisasi_teks(page["text"]) for page in pages)
```
